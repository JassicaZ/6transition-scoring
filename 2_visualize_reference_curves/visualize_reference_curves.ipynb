# 6-transition reference curve 

### The Z-score reference table for the age-6 transition is provided in this repository and can be loaded using:

```python
import pandas as pd

df = pd.read_csv("./data/reference_curve.csv")
df
```

### The following example shows how to visualize individual scores on the reference curve:

```python
# Load the transition scores from the previous step
df_score4JIA= pd.read_csv("../1_calculate_6transition_score/results/JIA/transition_score/transition_scores.csv")
```

```python
df_score4JIA
```

```python
import matplotlib.pyplot as plt

df=df[df['age']<=21]
x = df["age"]

# 创建画布
plt.figure(figsize=(8, 6))

# 依次画每条z-score曲线
for col in df.columns[1:]:
    plt.plot(x, df[col], label=f"z={col}")

#画个体点
plt.scatter(
    df_score4JIA["age"],
    df_score4JIA["6_transition_score"],
    color="black",
    s=20,
    alpha=0.6,
    label="Individuals from the JIA cohort"
)

# 设置标签
plt.xlabel("Age")
plt.ylabel("Score")
plt.legend(title="Z-score")
plt.tight_layout()

plt.show()
```
